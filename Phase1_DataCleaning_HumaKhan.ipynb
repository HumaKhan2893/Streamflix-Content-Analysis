Phase 1: Data Quality Report – StreamFlix Content Analytics Project

Data Quality Report

1. Introduction

A data quality assessment was conducted on the six StreamFlix datasets to evaluate their completeness, accuracy, consistency, and reliability before performing analysis. The datasets included subscribers.csv, titles.csv, watch_history.csv, watchlist.csv, ratings.csv, and reviews.csv. The assessment focused on identifying missing values, duplicate records, incorrect data types, outliers, and referential integrity issues.

2. Dataset Overview

All six datasets were successfully imported into Python using the Pandas library. The structure of each dataset was reviewed by examining the number of rows, columns, and data types. The datasets contained a combination of numerical, categorical, and date fields suitable for analytical processing.

3. Missing Values

Each dataset was checked for missing (null) values. Columns containing missing values were identified, and the percentage of affected records was calculated. Most datasets contained either no missing values or only a small percentage of missing data, indicating a high level of completeness. Any missing values identified will be handled during the data cleaning phase using appropriate techniques such as imputation or removal, depending on their impact on the analysis.

4. Duplicate Records

Duplicate rows were examined across all datasets. Special attention was given to the watch_history.csv and subscribers.csv datasets by checking for duplicate watch_id and subscriber_id values. No unexpected duplicate primary key values were identified, confirming that each record represents a unique entity. Where duplicate rows existed, they were flagged for removal during the cleaning process.

5. Data Type Validation

Data types were validated to ensure consistency with the data dictionary. Date fields such as watch_date and signup_date were converted from text format to proper datetime objects. Numerical fields including watch_duration_min, subscription prices, and rating values were verified as numeric data types. This ensures accurate calculations and time-based analysis in later project stages.

6. Outlier Detection

The watch_duration_min field was examined for unusual values using descriptive statistics and the Interquartile Range (IQR) method. Potential outliers, such as exceptionally long viewing sessions or watch durations exceeding the corresponding title length, were identified and flagged for further investigation. These records may indicate data entry errors or exceptional user behavior.

7. Referential Integrity

Relationships between datasets were validated by comparing foreign keys with their corresponding primary keys. All subscriber_id values in watch_history.csv were checked against subscribers.csv, and all title_id values were verified against titles.csv. This validation ensures that every viewing record references an existing subscriber and title, maintaining consistency across the database.

8. Overall Assessment

The StreamFlix datasets demonstrate a good level of data quality and are generally suitable for exploratory data analysis. Minor issues, such as missing values, duplicate records (if any), and potential outliers, have been identified for correction during the data cleaning phase. Following these cleaning steps, the datasets will be well-prepared for visualization, statistical analysis, and business intelligence reporting.

Conclusion

The Phase 1 data quality assessment confirms that the StreamFlix datasets provide a reliable foundation for analysis. Data profiling, validation, and integrity checks were successfully completed, and the identified issues have been documented for resolution in the next phase. This assessment ensures that subsequent analyses and business recommendations are based on accurate, consistent, and trustworthy data.
